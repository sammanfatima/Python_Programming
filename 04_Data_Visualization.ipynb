# **Data Visualization**

#### **Import Libraries**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

### **Import titanic data from seaborn library**

```python
df = sns.load_dataset('titanic')
```

```python
df.plot(kind='scatter', x='age', y='fare', color='green', figsize=(10,6))
```

```python
# change color using hex code
df.plot(kind='scatter', x='age', y='fare', color='#4c045c', figsize=(10,6))
```

```python
df.head()
```

```python
df.groupby('sex')['age'].mean()
```

```python
df.groupby('sex')['age'].mean().plot(kind='bar')
plt.ylabel('Average Fare')
plt.xlabel('Gender')
```

```python
df.groupby('sex')['fare'].mean()
```

```python
df.groupby('sex')['fare'].mean().plot(kind='bar')
```

### **Seaborn plots**

#### **Bar plots using seaborn**

```python
# 0 dead, 1 survived
sns.barplot(data=df, x='survived', y='fare', color='purple')
```

```python
sns.barplot(data=df, x='sex', y='fare', color='purple')
```

```python
sns.barplot(data=df, x='sex', y='fare', hue='class')
```

```python
sns.barplot(data=df, x='sex', y='age', hue='class')
```

#### **Boxplot**

```python
sns.boxplot(data=df, x='sex', y='age', hue='class')
```

```python
sns.scatterplot(data=df, x='fare', y='age', hue='class')
```

```python
sns.scatterplot(data=df, x='fare', y='age', hue='sex')
```

#### **Make scatter plot via plotly**

```python
import plotly.express as px
```

```python
fig = px.scatter(df, x='age', y='fare', color='class')
fig.show()
```
